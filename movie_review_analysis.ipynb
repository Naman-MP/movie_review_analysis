```python
from transformers import pipeline
import gradio as gr
from smolagents import CodeAgent, InferenceClientModel, DuckDuckGoSearchTool
```

```python
analizer = pipeline("sentiment-analysis")
```

```python
llm_engine = InferenceClientModel("Qwen/Qwen2.5-Coder-32B-Instruct")
search_tool = DuckDuckGoSearchTool()

agent = CodeAgent(
    tools=[search_tool],
    model=llm_engine,
    instructions=(
    "You are an expert movie critic app. When a user asks about a movie, "
        "if you do not know the details or if it is a recent release, use the web search tool "
        "to look up its plot, reviews, or reception. Then, summarize your critique "
        "based on that context.")
)
```

```python
def critique_with_search(user_query):
  response = agent.run(user_query)
  return response
```

```python
custom_ui = gr.Interface(
    fn=critique_with_search,
    inputs = gr.Textbox(label="Movie name and review" , placeholder="Enter the movie review"),
    outputs = gr.TextArea(label="analysis of the movie"),
    title="🍿 Movie Review Sentiment Analyzer",
    description="This app analyzes your input strictly as a movie review.",
    examples=[["The pacing was slow but the acting was phenomenal."], ["Waste of two hours."]]
)


if __name__=='__main__':
  custom_ui.launch(share=False)
```
